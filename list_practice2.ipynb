```python
#minimum & maximum
x = [9,2,3,4,1,6,7]
minimum = x[0];
maximum = x[-1]
for i in range(len(x)-1):
  if x[i] < minimum:
    minimum = x[i]
  if x[i]> maximum:
    maximum = x[i]
print('minimum value is' + ' ' + str(minimum))
print('maximum value is' + ' ' +str(maximum))
```

```python
#reverse elements in tuple
x = [(1,2),(3,4),(4,5)]
res = []
for i in x:
  res.append(tuple(reversed(i)))
print(res)
```

```python
#add corresponding elements in list
x = [(1,2),(3,4)]
y = [(5,6), (7,8)]
result = []
for a,b in zip(x,y):
  result.append((a[0]+b[0], a[1]+b[1]))
print(result)
```
