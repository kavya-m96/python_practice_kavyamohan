```python
mirror_dict = {
    'H': 'S',
    'E': 'V',
    'L': 'O',
    'W': 'D',
    'O': 'L',
    'R': 'I',
    'D': 'W'
}
input_string = 'HELLO WORLD'
output_string = ''
for i in input_string:
    if i in mirror_dict:
        output_string += mirror_dict[i]
    else:
        output_string += ' '
print(output_string)
```

```python
#create dictionary of maps each uppercase and lowercase to ascii equivalent
dictionary = {}
list_of_char = ['A','a','B','b']
for i in list_of_char:
    dictionary[i] = ord(i)
print(dictionary)
```

```python
#count vowels in given string and store count in dictionary
vowel = 'I am a vowel'
vowel_list = ['a','e','i','o','u']
dictionary = {}
for i in vowel:
    if i.lower() in vowel_list:
        dictionary[i] = vowel.count(i)
print(dictionary)
```

```python
input = ['listen', 'silent', 'enlist', 'hello', 'lemon', 'melon', 'debitcard', 'badcredit']
anagrams = {}
for i in input:
  key = ''.join(sorted(i))
  if key in anagrams:
   anagrams[key].append(i)
  else:
    anagrams[key] = [i]
print(list(anagrams.values()))
```

```python
#dictionary is a collection of key value pairs
#they contain unique keys and values can be of any data type
#dictionary comprehension is one line of creating dictionary with key value assignment
#dict.get(key) and dict[key] are both ways of getting the value based on key but if key is not present dict[key] throws keyerror whereas dict.get(key) returns None
#merge two dictionaries
x = {1:2,3:4}
y = {5:6,7:8}
x.update(y)
print(x)
z = {**x,**y} #if both contains same key then the latest dictionary values only retained
print(z)
#update values in dictionary
x[1] = 3 #if key not present it adds it
x.update({'marks':90})
print(x)
```

```python
#iterate over dictionary to access keys,values,key value pairs
x = {'name': 'kavya', 'age': 10, 'marks': 90}
for key in x:
    print(key)
for value in x.values():
    print(value)
for key,value in x.items():
    print(key,value)
#ways of deleting items from dictionary
x.popitem()
print(x)
x.pop('age')
print(x)
del x['name']
print(x)
```
