```python
#add even numbers to list and odd number to another list
startVal = 0
endVal = 151
x = []
y = []

while startVal >= 0 and startVal <=endVal:
     if startVal %2 == 0:
         x.append(startVal)
     else:
         y.append(startVal)
     startVal += 1

print('Even ordered list is :' + x)
print('Even ordered list is :' +y)
```
